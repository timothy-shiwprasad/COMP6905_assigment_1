## Objective Function
This objective functions is created by the combination of three componets mentioned in the problem.  
1. Expected Net Return: ENR(x<sub>S</sub>, x<sub>F</sub>, x<sub>T</sub>, x<sub>A</sub>) = 500x<sub>S</sub> + 640x<sub>F</sub> + 790x<sub>T</sub> + 180x<sub>A</sub>

2. Utility Connection assesment per acre: UCA(x<sub>S</sub>, x<sub>F</sub>, x<sub>T</sub>, x<sub>A</sub>) = 120x<sub>S</sub> + 160x<sub>F</sub> + 190x<sub>T</sub> + 260x<sub>A</sub>

3. Actual Utility Conection cost : AUC(x<sub>S</sub>, x<sub>F</sub>, x<sub>T</sub>, x<sub>A</sub>) = max{3000, UCA(x<sub>S</sub>, x<sub>F</sub>, x<sub>T</sub>, x<sub>A</sub>)}

Final Objective Function:

Actual Net Return : ENR(x<sub>S</sub>, x<sub>F</sub>, x<sub>T</sub>, x<sub>A</sub>) = 500x<sub>S</sub> + 640x<sub>F</sub> + 790x<sub>T</sub> + 180x<sub>A</sub> - AUC(x<sub>S</sub>, x<sub>F</sub>, x<sub>T</sub>, x<sub>A</sub>)

For Simplicity:
Let x = (x<sub>S</sub>, x<sub>F</sub>, x<sub>T</sub>, x<sub>A</sub>)

ANR(x) - AUC(x)

Objective: max ANR(x) - AUC(x) ,x >= 0

# Question B

We will replace the max function AUC(x) and define it as a constrain.

u = AUC(x)

## New Constrains
- u >= 3000
- u >= 120x<sub>S</sub> + 160x<sub>F</sub> + 190x<sub>T</sub> + 260x<sub>A</sub>

New Objective Function:
Z(x)  = 500x<sub>S</sub> + 640x<sub>F</sub> + 790x<sub>T</sub> + 180x<sub>A</sub> - u


## Constrains
- x<sub>S</sub> + x<sub>F</sub> + x<sub>T</sub> + x<sub>A</sub> <= 18
- 22x<sub>S</sub> + 30x<sub>F</sub> + 45x<sub>T</sub> + 70x<sub>A</sub> <= 750
- 2.0x<sub>S</sub> + 2.8x<sub>F</sub> + 3.6x<sub>T</sub> + 5.0x<sub>A</sub> <= 60
- 0.6x<sub>S</sub> - 0.4<sub>F</sub> - 0.4x<sub>T</sub> - 0.4x<sub>A</sub> >= 0
- 900x<sub>S</sub> + 1250x<sub>F</sub> + 1500x<sub>T</sub> + x<sub>A</sub> - AUC(x<sub>S</sub>, x<sub>F</sub>, x<sub>T</sub>, x<sub>A</sub>) <= 26000
- u >= 3000
- u >= 120x<sub>S</sub> + 160x<sub>F</sub> + 190x<sub>T</sub> + 260x<sub>A</sub>

### Reasoning
The constrain forces u to be both  at least 3000 and AUC(x). Since u is being subtracted, trhe model will chose a U value that is smallest and matches the 2 constrains. This in return would ensure that we are maximizing the Net return.



```julia
using Pkg; Pkg.add("JuMP"); Pkg.add("Clp")

using JuMP
using Clp

model = Model(Clp.Optimizer)

# Decision variables In acres
@variable(model, xS >= 0)
@variable(model, xF >= 0)
@variable(model, xT >= 0)
@variable(model, xA >= 0)

# Auxiliary variable for utility charge in thousands
@variable(model, u >= 0)

# Constraints
@constraint(model, land, xS + xF + xT + xA <= 18)
@constraint(model, elec, 22xS + 30xF + 45xT + 70xA <= 750)
@constraint(model, water, 2.0xS + 2.8xF + 3.6xT + 5.0xA <= 60)
@constraint(model, starter_share, 0.6xS - 0.4xF - 0.4xT - 0.4xA >= 0)
@constraint(model, budget, 900xS + 1250xF + 1500xT + 2200xA + u <= 26000)
@constraint(model, u_min, u >= 3000)
@constraint(model, u_assess, u >= 120xS + 160xF + 190xT + 260xA)

# Objective: maximize net return after paying utility charge
@objective(model, Max, 500xS + 640xF + 790xT + 1080xA - u)

optimize!(model)
```

```julia
# Optimal Acres for each development chosen to maximize Expected Net Return
xS_opt = value(xS)
xF_opt = value(xF)
xT_opt = value(xT)
xA_opt = value(xA)
u_opt  = value(u)
obj_opt = objective_value(model)

println("The Optimal Acres for each development type is as follows: Starter Cottage Development: $(xS_opt) , Family House Development : $(xF_opt) , Trainer Townhouse Development: $(xT_opt), Apartment Development : $(xA_opt)")

println("The maximum Expected Net Return (In Thousands): $(obj_opt)")
```

```julia
land_lhs = value(xS + xF + xT + xA)
elec_lhs = value(22xS + 30xF + 45xT + 70xA)
water_lhs = value(2.0xS + 2.8xF + 3.6xT + 5.0xA)
budget_lhs = value(900xS + 1250xF + 1500xT + 2200xA + u)

println("Contraints bindings at the Optimal")
println("Acres Constraints: $(land_lhs)")
println("Electricity Contraints: $(elec_lhs)")
println("Water Constraints: $(water_lhs)")
println("Budget Constrains: $budget_lhs")
```

# Interpretation of Shadow Prices
The Shadow Prices for the contraints are as followed:

- Land : 152.00
- Electricty: 0.0
- Water: 0.0
- Budget: 0.414

The shadow prices for electricity and water are 0.0 , since there is still additional resources avaliable for electricity and water (slack) an increase in the KW of electricy being genereated or a increase of gallons of water provided will not increase our objective of maximizing our Net Returns. We essentially have more than enough.

On the other hand the shadow prices for Land and Budget are 152 and 0.414 respectively. These two contraints are binding and based on their shadow prices we can expect an increase of 152 `($000)` to our net return for every increase unit increase of acre. The same for the budget. A single  increase of  1 `($000)`  to the budget will result in an 0.414 `($000)` increase to our Net Return.
